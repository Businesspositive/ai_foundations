Exercise 4 - Performance Check

Compare a Python loop with numpy vectorized operation

```python
import time

nums = np.arange(1000000)

start_time = time.time()
sum_loop = [x**2 for x in nums]
print("Loop time: %s seconds" % (time.time() - start_time))

start_time = time.time()
sum_vectorized = np.sum(nums**2)
print("Vectorized time: %s seconds" % (time.time() - start_time))
```
