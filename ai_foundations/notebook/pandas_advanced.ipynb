Advanced Pandas (Data Handling for AI Systems)

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")
df.head()
```

Mini Exercises

Exercise 1: Filtering and Sorting

How can you find the average fare for male passengers older than 50?

```python
df[df['age'] > 50].sort_values('fare', ascending=False).head()
```

Exercise 2: Group By and Aggregation

```python
df.groupby(['class', 'sex'])['fare'].agg(['mean', 'max', 'min'])
```

Exercise 3 - Missing Data

```python
df.isna().sum()
```

Exercise 4 - Apply and Feature Enginering

```python
df['family_size'] = df['sibsp'] + df['parch'] + 1
df['fare_per_person'] = df['fare'] / df['family_size']
df[['family_size', 'fare_per_person']].head()
```
