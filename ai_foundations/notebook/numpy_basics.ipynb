🧩 NumPy Arrays in Detail

```python
import numpy as np
arr = np.array([[1, 2, 3,], [4, 5, 6]])
print(arr.shape) #print the shape of the array
print(arr.ndim) #print the number of dimensions
print(arr.dtype) #print the data type of the array elements
```

🧮 NumPy Math — More Than It Looks

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b) # element-wise addition
print(a * b) # element-wise multiplication
print(a.dot(b)) # dot product of two arrays like multiplication element-wise and then sum
```

```python
np.zeros((3, 3)) # Create a 3x3 array of zeros
```

```python
np.ones((2, 4)) # Create a 2x4 array of ones
```

```python
np.arange(0, 10, 2) # Create an array with values from 0 to 10 with a step of 2
```

```python
np.linspace(0, 1, 5) # Create an array of 5 values evenly spaced between 0 and 1
```

```python
np.random.rand(3, 3) # Create a 3x3 array of random values between 0 and 1
```

Challenge : Create a 3×3 NumPy array filled with random numbers between 0 and 1,
then normalize it — scale all numbers between 0 and 1 based on the maximum value

```python
arr = np.random.rand(3,3)
max_value = np.max(arr) or arr.max()
normalized_arr = arr / max_value
print("Original array:\n", arr)
print("Normalized array:\n", normalized_arr)
```
