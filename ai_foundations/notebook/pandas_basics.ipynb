⚙️ Detecting & Fixing Missing / Corrupted Data

```python
import pandas as pd
import numpy as np

data = {
    'Date': ['2025-10-01', '2025-10-02', '2025-10-03', '2025-10-04'],
    'Company': ['TATAMOTORS', 'INFOSYS', 'RELIANCE', 'HDFC'],
    'Price': [894.5, np.nan, 1563.0, 1723.2],
    'Volume': [1234567, 1278900, '-', 1043300],
    'Sector': ['Auto', 'Auto', 'Energy', 'Finance'],
    'Rating': ['A', 'B', np.nan, 'A']
}

df = pd.DataFrame(data)
```

```python
df.head(2)
```

```python
df.tail(2)
```

1️⃣ Checking for missing or invalid data

```python
print(df.info())
```

```python
print(df.isna())
```

2️⃣ Fixing the Volume column

We’ll convert it to numeric and tell Pandas to coerce errors into NaN:

```python
df['Volume'] = pd.to_numeric(df['Volume'], errors='coerce')
df.head()
```

3️⃣ Handling Missing Data

We can fill or drop them:

```python
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
df['Price']=df['Price'].fillna(df['Price'].mean())
df['Rating']=df['Rating'].fillna('B')
print(df)
```

🧩 Why AI engineers care:
ML models can’t handle missing or string-type data in numeric columns — this cleaning step ensures data integrity.

🧮 Feature Engineering — Making Data Smarter

Feature engineering = creating new useful features from existing ones.

Example:

```python
# convert 'Date' column to datetime format
df['Date'] = pd.to_datetime(df['Date'])

#create new features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day

#add derived matrix price * volume

df['trade_value'] = df['Price'] * df['Volume']
print(df)
```
